## 1) 3D QRAFIKININ TEHLILI  Riyazi modelin komputer Interpretatasiyasi 
## (Relyef:Funksiyanin relyefinin coxlu sayda lokal maxsimum (tepeler)ve lokal minumumlar (cokeklukler)aydin gorunur.
## cetinlikler:Bele relyefe malik funksiyalarda qlobal minumumu tapmaq cetin oldugu ucun xususi optimallasdirma alqoritminden istifade olunur

## 2)   2D Qrafikinin tehlili ve Vizual Tesdiq  Istilik xeritesi , Minumum noqtesi 

```python
import numpy as np                  # Riyazi hesablamalar və massivlər üçün
from scipy import optimize          # Minimum nöqtəni tapmaq (optimallaşdırma) üçün
import matplotlib.pyplot as plt    # Qrafikləri çəkmək üçün
from mpl_toolkits.mplot3d import Axes3D  # 3D qrafik dəstəyi üçün

```

```python
# İkidəyişənli funksiyanı təyin edən Python funksiyası:
def altibel(x):
    return ((4 - 2.1*x[0]**2 + x[0]**4 / 3.) * x[0]**2 + x[0] * x[1] \
            + (-4 + 4*x[1]**2) * x[1] **2)

# Şəbəkənin yaradılması:
x = np.linspace(-2, 2)
y = np.linspace(-1, 1)
xg, yg = np.meshgrid(x, y)

# 3D qrafikin çəkilməsi:
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(xg, yg, altibel([xg, yg]), rstride=1, cstride=1,
                       cmap=plt.cm.jet, linewidth=0, antialiased=False)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x, y)')
ax.set_title('Altı sayda Dəvə-belli funksiya')
plt.show()

# Minimumun hesablanması:
x_min = optimize.minimize(altibel, x0=[0, 0])
print('Funksiyanın optimallaşdırmasının nəticələri: ')
print(x_min)

# 2D qrafik və minimum nöqtəsinin göstərilməsi:
fig = plt.figure()
plt.imshow(altibel([xg, yg]), extent=[-2, 2, -1, 1], cmap='cool', origin="lower")
plt.colorbar()
plt.scatter(x_min.x[0], x_min.x[1], color='r', s=100, zorder=5) # Minimum nöqtəni qırmızı nöqtə ilə işarələyirik
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
