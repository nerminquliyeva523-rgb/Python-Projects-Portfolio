```python
%%writefile ilk_saytam.py
import streamlit as st

# Saytın brauzer tırnağındakı adını və ikonunu nizamlayırıq (Ən üstdə olmalıdır)
st.set_page_config(page_title="Basic Academy | Python Trainer", page_icon="💻", layout="wide")

# ---- SOL MENYU (SIDEBAR) ----
with st.sidebar:
    st.image("https://cdn-icons-png.flaticon.com/512/5968/5968350.png", width=80) # Python İkonu
    st.title("İdarəetmə Paneli")
    st.write("Saytın funksiyalarını buradan nizamlayın:")
    
    # Giriş elementlərini sol tərəfə daşıyırıq ki, ana səhifə təmiz qalsın
    ad = st.text_input("📝 Adınızı daxil edin:")
    secilen_reng = st.color_picker("🎨 Uğurlu rənginizi seçin:", "#00b4f9")
    yas = st.slider("🎂 Yaşınız:", min_value=1, max_value=100, value=25)
    
    st.markdown("---")
    st.caption("© 2026 Basic Academy. Bütün hüquqlar qorunur.")

# ---- ANA SƏHİFƏ ----

# Başlıq hissəsi
st.title("💻 Build Your Future with Informatics")
st.subheader("Python Dünyasına Xoş Gəldiniz! 🚀")
st.write("Bu platforma Python proqramlaşdırma dilini və müasir backend texnologiyalarını interaktiv şəkildə öyrənmək üçün yaradılmışdır.")

st.markdown("---")

# Tablar (Səhifəni hissələrə bölmək üçün möhtəşəm dizayn aləti)
tab1, tab2, tab3 = st.tabs(["👋 Qarşılanma", "📚 Dərs Materialı", "🧠 Kiçik Sınaq"])

with tab1:
    st.markdown("### 🌟 İnteraktiv Zona")
    st.write("Sol menyudan məlumatlarınızı daxil edib aşağıdakı düyməni sıxa bilərsiniz.")
    
    if st.button("✨ Məni Salamla"):
        if ad:
            # Seçilən rəngi vizual olaraq xəbərdarlıq blokunda istifadə edirik
            st.success(f"Xoş gəldiniz, {ad}! Sizin {yas} yaşınız var. 🎉")
            st.markdown(f"**Seçdiyiniz rəng:** <span style='color:{secilen_reng}; font-weight:bold;'>{secilen_reng}</span>", unsafe_allow_html=True)
        else:
            st.warning("Zəhmət olmasa, sol menyudan adınızı daxil edin! ⚠️")

with tab2:
    st.markdown("### 📖 Python-da `def` Funksiyaları")
    st.write("Gəlin təmiz kod yazma standartlarına uyğun olaraq ilk funksiyamıza baxaq:")
    
    # Professional kod bloku göstəricisi
    kod = """
def salamla(istifadeci_adi):
    print(f"Salam, {istifadeci_adi}! Kodlaşdırmaya xoş gəlmisən.")

# Funksiyanı çağırmaq
salamla("Nermin")
    """
    st.code(kod, language="python")
    st.info("💡 Qeyd: Clean Code standartları üçün dəyişən və funksiya adlarını ingilis dilində (`user_name`, `greet`) təyin etmək sənaye təcrübəsində daha çox üstünlük təşkil edir.")

with tab3:
    st.markdown("### 📝 Günün Sınaq Sualı")
    st.write("**Sual:** Aşağıdakı Python kodunun nəticəsi nə olacaq?")
    st.code('print("Python"[::2])', language="python")
    
    # Radio button ilə test variantları
    cavab = st.radio(
        "Düzgün variantı seçin:",
        ["A) Python", "B) Pto", "C) yhn", "D) Pth"],
        index=None,
        placeholder="Variant seçin..."
    )
    
    if st.button("Cavabı Yoxla 🎯"):
        if cavab == "B) Pto":
            st.balloons() # Ekrana şarlar uçurur 😍
            st.success("Təbriklər! Düzgün cavab yazmısınız. `[::2]` addım parametri simvolları 2-2 atlayaraq götürür.")
        elif cavab is None:
            st.error("Zəhmət olmasa, əvvəlcə bir variant seçin!")
        else:
            st.error("Yanlış cavab! Bir daha cəhd edin. 💡 İpucu: İndekslərə diqqət yetirin (0, 2, 4).")
```
