```python
#Shallow Copy (Sethi nusxeleme) # Shallowda orginal deyisir 
orginal={'a':1,'b':[1,2]}
copied=orginal.copy()   #kopyasin cixartdi
copied['a']=100
copied['b'].append(9)  #siyanin sonuna 3 elave etdi
print(orginal)
```

```python
# Deep Copy (Derin nusxeleme) Orjinal deyismir
import copy
orginal={'a':1,'b':[1,2]}
deep_copied=copy.deepcopy(orginal)
deep_copied['b'].append(99)
print(orginal)
```

```python
#Acar ve deyerlerin yerini deyismek(Invert dict)
orginal={'a':1,'b':2,'c':3}
inverted={value:key for key ,value in orginal .items()}
print(inverted)
```

```python
#Shertli (if) dictionary Comprehension
numbers={'a':1,'b':2,'c':3,'d':4}
even_numbers={k: v for k,v in numbers.items() if v%2==0}
print(even_numbers)
```

```python
#Serli if-else ile deyer deyisdirme
scores={'Ali':45,'Vali':75,'Sara':90}
results={k:('Kecdi' if v>=50 else 'Kesildi')for k,v in scores.items()}
print(results)
```

```python
#Metindeki herslerin tezliyini hesablamaq
text='python'
char_count={char:text.count(char) for char in set(text)}
print(char_count)
```

```python
#Standart collections.defaultdict islenmesi

from collections import defaultdict
groups=defaultdict(list)
pairs=[('fruit','apple'),('veg','carrot'),('fruit','banan')]
for category,item in pairs:
    groups[category].append(item)
print(dict(groups))
```

```python
#Acarlari boyuk herife cevirmek 
raw_data={'name':'Eldar','city':'Baku'}
clean_data={k.capitalize():v for k ,v in raw_data.items()}
print(clean_data)
```

```python
#Lugetden None deyerleri filtrlemek
data={'a':10,'b':None,'c':30,'d':None}
filtred={k:v for k,v  in data.items() if v is not None }
print(filtred)
```

```python
#Temperatorlari Selside Farenheyte cevirmek 
celsius={'Baku':25,'London':15,'Moscow':5}
fahrenheit={city:(temp*9/5)+32 for city,temp in celsius.items()}
print(fahrenheit)
```
